## 3. Check the GPU

```
!nvidia-smi --query-gpu=name,driver_version,memory.total --format=csv
import numba.cuda as cuda, cupy as cp
if not cuda.is_available():
    raise SystemError('No GPU! Runtime > Change runtime type > T4 GPU')
gpu = cuda.get_current_device()
name = gpu.name.decode() if isinstance(gpu.name, bytes) else gpu.name
print('GPU               :', name)
print('Compute capability:', gpu.compute_capability)
print('Warp size         :', gpu.WARP_SIZE)
print('Max threads/block :', gpu.MAX_THREADS_PER_BLOCK)
print('Shared mem/block  :', gpu.MAX_SHARED_MEMORY_PER_BLOCK/1024, 'KB')
```
